## #02. 데이터 전처리
### 1. 명목형 라벨링
- 명목형 컬럼(CHAS)은 category 타입이라 그대로는 모델에 넣을 수 없으므로 정수로 바꾼다.

```python
df1 = my_prep.labeling(df, columns=nominal_cols)
df1.head()
```

### 2. 더미변수 인코딩
- 라벨링된 명목형 컬럼을 더미변수로 펼친다.
    - 이 과정까지는 모델에 넣기 위한 필수 변환이므로, 이 시점의 데이터프레임을 기준선으로 저장한다.

```python
df2 = my_prep.dummies(df1, columns=nominal_cols, drop_first=True)
df2.to_excel("체크포인트0_기준선.xlsx", index=False)
df2.head()
```

### 3. 로그 변환 대상 컬럼 선정

```python
# 우측꼬리 로그 변환 대상 컬럼
log_cols = desc.loc[desc['log_need']=='log1p'].index.tolist()
log_cols
```

```python
# 좌측꼬리 로그 변환 대상 컬럼
reflect_cols = desc.loc[desc['log_need']=='reverse_log1p'].index.tolist()
reflect_cols
```

### 4. 로그 변환

```python
df3 = my_prep.log_transform(df2, log_columns=log_cols, reflect_columns=reflect_cols)
df3.to_excel("체크포인트1_로그변환.xlsx", index=False)
df3.head()
```

### 5. 이상치 대체
- IQR 울타리('Q1 - 1.5 * IQR Q3 + 1.5 * IQR')를 벗어난 값을 경계값으로 대체한다.
    - 행을 삭제하지 않으므로 데이터 수(506개)가 그대로 유지된다.
- 경계값('bound')을 쓰는 이유는 값의 순서가 보존되기 때문이다.
    - 중앙값으로 대체하면 범죄율이 가장 높은 동네가 가장 안전한 동네로 뒤바뀌어 버린다.
- 앞에서 로그를 취해둔 덕분에 잘려나가는 값이 크게 줄어든다.
    - CRIM을 보면 원본 기준 이상치가 66개(13.0%)였는데, 로그 변환 후에는 6개(1.2%)만 울타리를 벗어난다.

```python
df4 = my_prep.replace_outlier(df3, columns=feature_continuous)
df4.to_excel("체크포인트2_이상치대체.xlsx", index=False)
df4.head()
```

### 6. 다중 공선성 제거
- VIF가 임계값을 넘는 변수를 하나씩 제거한다.
- 종속변수(MEDV)는 대상에서 빼야하므로 연속형 독립변수만 지정한다.

```python
df5 = my_prep.reduce_vif(df4, columns=feature_continuous, threshold=10.0)
df5.to_excel("체크포인트3_다중공선성제거.xlsx", index=False)
df5.head()
```

### 7. 정규화
- 연속형 독립변수를 대상으로 한다.
- 정규화는 OLS의 R2를 바꾸지 못한다.
    - 각 변수에 상수를 더하고 곱하는 선형 변환일 뿐이고, 회귀계수가 그 반대로 조정되면서 예측값이 완전히 동일해지기 때문이다.
    - Ridgh, Lasso처럼 계수에 벌점을 주는 모델을 쓸 때다.

```python
df6 = my_prep.scaling(df5, columns=feature_continuous, method='standard')
df6.to_excel("체크포인트4_정규화.xlsx", index=False)
df6.head()
```
