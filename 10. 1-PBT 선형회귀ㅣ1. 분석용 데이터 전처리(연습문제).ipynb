```python
from hossam import load_data
from helpers import *
```

```python
origin = load_data('insurance_qtcheck')
origin.head()
```

```python
desc = load_data('insurance_qtcheck_desc')
desc.T
```

```python
cat_desc = load_data('insurance_qtcheck_category_desc')
cat_desc
```

```python
df = my_qtcheck.set_type(origin, as_category=['sex', 'smoker', 'region'])
```

```python
# 종속변수 이름
target = "charges"
# 명목형 변수 이름
nominal_cols = my_qtcheck.get_categorical_column_names(df)
# 연속형(종속변수 포함)
continuous_cols = my_qtcheck.get_number_column_names(df)
# 연속형 독립변수
feature_continuous= [c for c in continuous_cols if c != target]
```

```python
print("목표변수:", target)
print("연속형(종속변수 포함) :", continuous_cols)
print("독립변수(명목형) :", nominal_cols)
print("독립변수(연속형) :", feature_continuous)

display(df.head())
display(desc)
display(cat_desc)
```

```python
df1 = my_prep.labeling(df, columns=nominal_cols)
df1.head()
```

```python
df2 = my_prep.dummies(df1, columns=nominal_cols, drop_first=True)
df2.to_excel("10.연습문제_체크포인트0_기준선.xlsx", index=False)
df2.head()
```

```python
# 우측꼬리 로그 변환 대상 컬럼
log_cols = desc.loc[desc['log_need']=='log1p'].index.tolist()
log_cols
```

```python
df3 = my_prep.log_transform(df2, log_columns=log_cols, reflect_columns=reflect_cols)
df3.to_excel("10.연습문제_체크포인트1_로그변환.xlsx", index=False)
df3.head()
```

```python
df4 = my_prep.replace_outlier(df3, columns=feature_continuous)
df4.to_excel("10.연습문제_체크포인트2_이상치대체.xlsx", index=False)
df4.head()
```

```python
df5 = my_prep.reduce_vif(df4, columns=feature_continuous, threshold=10.0)
df5.to_excel("10.연습문제_체크포인트3_다중공선성제거.xlsx", index=False)
df5.head()
```

```python
df6 = my_prep.scaling(df5, columns=feature_continuous, method='standard')
df6.to_excel("10.연습문제_체크포인트4_정규화.xlsx", index=False)
df6.head()
```
