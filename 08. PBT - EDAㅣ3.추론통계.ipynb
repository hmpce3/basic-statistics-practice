## #03. T검정
### 1. 이진 명목형 변수 ↔ 연속
- CHAS(강 인접)에 따른 MEDV 차이
    - 검정값을 wide 포맷(두 컬럼)으로 분리해 전달한다.

```python
chas_medv = DataFrame({
    "비인접(0)": df.loc[df['CHAS'] ==0, target].reset_index(drop=True),
    "인접(1)": df.loc[df['CHAS'] ==1, target].reset_index(drop=True)
})

my_stats.test_independent(chas_medv, "비인접(0)", "인접(1)", title='CHAS에 따른 MEDV 차이')
```

- 인사이트
- 명목(2) X 연속 → T검정 계열, 정규성 위배 →Mann-Whitney U(비모수) 자동 적용

|항목|결과
|-|-|
|검정|Mann-Whitney U(비모수)|
|U 통계량|5605.5|
|p-value|0.002(유의)|
|방향(단측)|비인접(0) < 인접(1), p=0.001|
|결론|강에 인접한 지역(CHAS=1)의 MEDV가 더 높다|

- 해석 : 찰스강 인접 여부에 따라 집값이 통계적으로 유의하게 다르며, 인접지역이 더 비쌈
- p<0.05 유의, → CHAS를 독립변수 후보로 채택
- 비모수 검정이라 Cohen's d가 출력되지 않음 → 실질 크기는 박스플롯의 중앙값 차이로 보완판단

## #04. ANOVA
### 1. 3개 이상의 그룹을 갖는 명목형 파생변수 생성
- 원본엔 3집단 + 명목형이 없기 때문에 가상의 파생변수를 생성해서 진행
- 이 검정의 결과는 최종 변수 선별에는 포함되지 않음

```python
df['RM_grp']=cut(df['RM'], 3, labels=['방적음', '방보통', '방많음'])
df['RM_grp'].value_counts().sort_index()
```

### 2. 일원분산분석

```python
my_stats.anova_oneway(df, y='MEDV', between='RM_grp')
```

- 인사이트
- 명목(3+) x 연속 → ANOVA. RM을 3구간(방적음/방보통/방많음)으로 나눈 파생변수로 검정. 등분산 위배 → Welch_anova 자동적용

|항목|결과|
|-|-|
|검정|Welch_anova(등분산 위배)|
|F|125.757|
|p-value|<0.001 → 유의|
|효과크기(np2)|0.428 → Large(>=0.14)|
|그룹 크기|방적음 24/방보통 422/방많음 60(불균형)|

- 해석 : 방 개수 구간에 따라 MEDV가 통계적으로 유의하게 다르며, 효과크기도 매우 큼
- F만으로는 '어느 쌍'이 다른지 알 수 없음 > 사후 검정 필요

### 2. 사후 검정

```python
my_stats.posthoc_oneway(df, y='MEDV', between='RM_grp', title='RM구간별 MEDV 사후검정')
```

- 인사이트
- 정규성, 등분산 위배 → 사후검정은 Games-Howell	적용, 세쌍 모두 유의

|비교쌍|평균 A→B|차이|p|효과크기(hedges)|
|-|-|-|-|-|
|방적음vs방보통|-5.310|0.024(*)|-0.789|
|방적음vs방많음|-23.295|<0.001|-2.645|
|방보통vs방많음|-17.985|<0.001|-2.620|

- 방향 : 방적음(15.34)<방보통(20.65)<방많음(38.64) → 방이 많을수록 집값이 뚜렷하게 상승
- 모든 쌍이 유의하며, 특히 방많음 그룹이 다른 두 그룹과 매우 큰 차이를 보임
- 방 개수가 집값을 단조 증가시키는 강한 요인임을 사후검정이 재확인
